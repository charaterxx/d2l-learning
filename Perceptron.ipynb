### 感知机的简单实现

```python

def AND(x1 , x2):
    w1 , w2 , theta = 0.5 ,0.5 , 0.7
    tmp = x1 * w1 + x2 * w2
    if tmp <= theta:
        return 0
    else:
        return 1
AND(0,0),AND(0,1),AND(1,0),AND(1,1)
```

### 权重和偏置实现

```python
def AND(x1 , x2):
    x = np.array([x1 , x2])
    w = np.array([0.5 , 0.5])
    b = -0.7
    tmp = np.sum(w*x) + b
    if tmp <= 0:
        return 0
    else:
        return 1
```

### 与非门 或门

```python
def NAND(x1 , x2):
    x = np.array([x1 , x2])
    w = np.array([-0.5 , -0.5])
    b = 0.7
    tmp = np.sum(w*x) + b
    if tmp <= 0:
        return 0
    else:
        return 1
    
def OR(x1 , x2):
    x = np.array([x1 , x2])
    w = np.array([0.5 , 0.5])
    b = -0.2
    tmp = np.sum(w*x) + b
    if tmp <= 0:
        return 0
    else:
        return 1
```

### 感知机的局限性
**无法实现异或门**
想象一个坐标，(0,0) (0,1) (1,0) (1,1)分别是0 1 1 0，不存在一个直线能把0和1的结果分开。
但是把**直线的限制条件去掉**就可以用感知机表示。
曲线分割的成为非线性空间，直线分割的成为线性空间。

也可以用电路的组合，用与门、与非门、、或门进行配置，组合出来一个异或门


```python
# 异或门
def XOR(x1 , x2):
    s1 = NAND(x1 , x2)
    s2 = OR(x1 , x2)
    y = AND(s1 , s2)
    return y

XOR(0,0),XOR(0,1),XOR(1,0),XOR(1,1)
```

与门、或门是单层感知机
异或门是2层感知机
叠加了多层感知机也称为多层感知机


### 从与非门到计算机
计算机是处理信息的机器，输入一些内容，再进行特定的处理，然后输出结果。
这个很像感知机
（严格来说是激活函数使用了非线性的sigmoid函数的感知机）2层感知机可以表示任意函数

感知机内容
1.感知机是一个有输入输出的算法
2.感知机将权重和偏置设定为参数
3.使用感知机可以表示逻辑电路
4.异或门需要2层
5.单层感知机只能表示线性空间，多层感知机可以表示非线性空间
6.多层感知机从理论上可以表示计算机
